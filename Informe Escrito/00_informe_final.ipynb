## **Introducción**
**Contexto**

El análisis de datos en el mercado inmobiliario de alquiler es fundamental en la industria de bienes raíces, ya que influye en las decisiones de arrendadores, inquilinos, inversores y autoridades urbanas.

El conjunto de datos “Apartment for Rent Classified” proviene del repositorio UCI Machine Learning y contiene anuncios clasificados de apartamentos en alquiler en Estados Unidos, con atributos como precio, área (pies cuadrados), amenidades (p. ej., internet, piscina) y ubicación geográfica.

**Problema general y relevancia**

Detectar, limpiar y transformar los datos de anuncios para permitir tareas analíticas y predictivas (por ejemplo, modelar el precio de renta). Esto es relevante porque:

Mejora la transparencia del mercado al producir estimadores robustos.

Permite servicios de recomendación y valoración automática.

Reduce sesgos debidos a datos sucios (anuncios incompletos o duplicados)


**Alcance de esta primera entrega**

## **Planteamiento del problema**

Pregunta de negocio: predecir el precio de alquiler de un apartamento usando atributos del anuncio (ubicación, tamaño, recámaras/baños, amenidades, etc.) y, de forma complementaria, clasificar anuncios (ej. rango de precios o tipo de pago).

Variables clave: entradas como bedrooms, bathrooms, square_feet, amenities, cityname, state, pets_allowed; variable objetivo principal: price (USD). Tareas adicionales: clasificación con price_type o pets_allowed.

Impacto esperado: facilitar precios justos, ayudar a inquilinos y dueños en la toma de decisiones, y aportar insights sobre tendencias de mercado y asequibilidad.

Sesgo de muestreo: los datos provienen de anuncios en portales específicos, por lo que pueden no reflejar todo el mercado (ejemplo: ausencia de zonas rurales o precios atípicos).

Variables textuales complejas: atributos como amenities, title o body contienen muchos valores posibles, lo que dificulta su codificación sin pérdida de información.

Data leakage: se deben excluir variables redundantes, como price_display (derivada directamente de price), para evitar filtrar la respuesta.

Distribuciones sesgadas: es probable encontrar outliers de precio asociados a ciertas zonas o amenidades, lo que requerirá normalización o modelos robustos.

Limitaciones temporales: aunque existe la variable de fecha de publicación, en esta etapa no se abordarán dinámicas temporales avanzadas (forecasting).

Calidad de datos: pese a que no hay valores faltantes, pueden presentarse inconsistencias de formato (ejemplo: símbolos o comas en los precios) que requerirán limpieza.

### **Preprocesamiento y División de Datos**

El conjunto de datos fue sometido a un proceso de limpieza y normalización que garantizó coherencia y comparabilidad entre características. Las variables numéricas fueron imputadas mediante la mediana del conjunto de entrenamiento, mientras que las categóricas se transformaron mediante codificación one-hot utilizando un esquema que permite ignorar categorías no vistas durante el entrenamiento, evitando inconsistencias en validación y prueba. Además, las variables numéricas fueron escaladas con StandardScaler, técnica que favorece la convergencia y estabilidad en modelos sensibles a la magnitud de los atributos.

Los datos se dividieron en entrenamiento (70%), validación (15%) y prueba (15%), asegurando que la evaluación del desempeño general fuera objetiva y no estuviera influenciada por el proceso de optimización.

# **Entrenamiento y Registro de Experimentos (MLflow+ Databricks)**

En este proyecto se evaluaron distintos enfoques de modelado para predecir el precio de renta utilizando información estructurada del inmueble, su ubicación y características asociadas. Se seleccionaron cuatro modelos representativos del espectro de técnicas empleadas en problemas de regresión tabular: Regresión Lineal, Random Forest, XGBoost y LightGBM. La elección de estos modelos responde al objetivo de comparar métodos interpretables, robustos y de alto rendimiento dentro de un marco experimental reproducible.

La Regresión Lineal se incluyó como referencia base, debido a su simplicidad e interpretabilidad, permitiendo evaluar si existía una relación lineal suficiente entre las variables del inmueble y el precio. Por su parte, Random Forest ofrece una alternativa no lineal, robusta ante valores atípicos y menos sensible al preprocesamiento, aportando estabilidad y un desempeño consistente en datos heterogéneos. Finalmente, XGBoost y LightGBM representan técnicas avanzadas basadas en boosting, conocidas por alcanzar altos niveles de precisión en datos tabulares al capturar relaciones complejas e interacciones entre variables.

### **Selección al mejor modelo** 

Los modelos basados en árboles —Random Forest, XGBoost y LightGBM— fueron sometidos a un proceso de optimización sistemática mediante Optuna, utilizando el algoritmo TPE (Tree-structured Parzen Estimator). El objetivo de la optimización fue minimizar el RMSE en el conjunto de validación, considerando intervalos amplios de hiperparámetros que permitieran explorar distintos niveles de profundidad, regularización y complejidad del modelo.

Cada prueba generada por Optuna quedó registrada dentro del entorno de MLflow corriendo en Databricks, incluyendo métricas, parámetros evaluados y artefactos generados. Esto permitió mantener un historial completo de experimentos, facilitar la comparación entre modelos y asegurar la trazabilidad de los resultados.


Todos los experimentos se almacenaron en un experimento central dentro de Databricks. Para cada modelo se registraron:

Métricas de desempeño (RMSE, MAE y R²).

Parámetros utilizados en el entrenamiento.

El modelo resultante en formato MLflow Model.

Los artefactos del preprocesamiento (encoder y scaler) empleados para reproducir el flujo completo.

Las capturas incorporadas abajo muestran la visualización generada por Databricks, incluyendo el ID del experimento y del run correspondiente al mejor modelo según la métrica objetivo. Este registro garantiza la reproducibilidad del proceso y permite ejecutar nuevamente el modelo con las mismas condiciones en cualquier momento.

### **Orquestation (Training Pipeline)**

El pipeline incorpora una fase de optimización de hiperparámetros basada en Optuna, donde se entrenan iterativamente modelos Random Forest, XGBoost y LightGBM. Cada ensayo se ejecuta dentro de un contexto de MLflow, registrando parámetros, métricas y artefactos relevantes. Esta etapa permite seleccionar automáticamente los mejores conjuntos de hiperparámetros minimizando el RMSE. A partir de los mejores parámetros encontrados, el flujo avanza hacia la fase de Train Best Models, donde cada modelo es reentrenado con su configuración óptima, registrando nuevamente sus métricas finales en MLflow. Finalmente, el pipeline ejecuta un proceso de selección Champion/Challenger que identifica los modelos con mejor desempeño según la métrica configurada (R² en este caso), y registra ambos en el Model Registry con sus respectivos alias, facilitando la gestión de versiones y una estrategia sistemática de comparación entre modelos.

La ejecución completa está encapsulada en el flujo principal Proyecto-Final-Precios-Flow, que establece la conexión con MLflow en Databricks, carga variables de entorno mediante dotenv, define el experimento y orquesta todas las tareas mencionadas en orden secuencial. Su ejecución puede iniciarse simplemente llamando al script desde la terminal o mediante agentes de Prefect. El flujo está preparado para registrar logs detallados de cada etapa y, gracias a las capacidades nativas de Prefect, permite manejar fallos mediante reintentos automáticos, seguimiento en tiempo real y la posibilidad de configurar alertas ante eventos de error o interrupción del pipeline, garantizando así una operación robusta y fácilmente monitoreable.

### **Servir el modelo (API FastAPI)**

Este código crea un sistema completo para hacer predicciones de precios usando un modelo entrenado previamente. Funciona así: FastAPI sirve como la “API” que recibe los datos, los procesa y devuelve la predicción del modelo, mientras que Streamlit es la interfaz bonita donde el usuario llena un formulario y ve el resultado.

Primero, el código carga algunas cosas importantes del entrenamiento: el encoder, el scaler y la lista de columnas originales, que están guardadas como archivos pickle. También se conecta a MLflow (que está en Databricks) y busca cuál es el mejor modelo que quedó registrado como champion. Ese modelo se carga automáticamente para usarlo en producción.

Después, hay una función de preprocesamiento que toma los datos que el usuario envía, los convierte en un DataFrame y les hace exactamente las mismas transformaciones que se usaron cuando el modelo se entrenó: codifica las columnas categóricas con One-Hot, ordena todas las columnas para que coincidan con las del entrenamiento, llena las faltantes y escala los valores. Esto es importantísimo para que el modelo no falle.

Luego está la función hacer_predic, que simplemente toma esos datos procesados y llama al modelo campeón para que haga la predicción final.

La parte de FastAPI define una ruta POST llamada /predict. Cuando un usuario envía datos ahí, la API llama a las funciones anteriores, calcula la predicción y devuelve el resultado en formato JSON.

La última parte del código es la app de Streamlit. Esta crea una interfaz gráfica donde las personas pueden escribir los valores: categoría, baños, metros cuadrados, latitud, etc. Cuando se hace clic en “Predecir”, la app arma un JSON con todos esos valores y lo manda a la API de FastAPI. Cuando la API responde, Streamlit muestra el valor predicho.

## **Conclusiones**
El proyecto logró implementar un flujo completo de machine learning combinando preprocesamiento estructurado, optimización automatizada de modelos y un sistema formal de gestión de experimentos. El uso de técnicas como imputación, codificación categórica con One-Hot Encoding y escalamiento garantizó que todos los modelos trabajaran con datos consistentes y preparados de forma adecuada. Además, la integración de algoritmos lineales, modelos basados en árboles y métodos avanzados de boosting permitió analizar el problema desde diferentes enfoques y entender mejor las relaciones presentes en el dataset.

La selección del mejor modelo se realizó con criterios cuantitativos, evaluando métricas clave como MAE, RMSE y R². Después de comparar los resultados, XGBoost mostró un desempeño superior, menor error de predicción y mayor estabilidad entre los conjuntos de entrenamiento, validación y prueba. Esto lo convirtió en la opción más sólida tanto desde el punto de vista predictivo como operativo. El modelo final fue registrado en el Model Registry de MLflow como champion, asegurando control de versiones, trazabilidad y facilidad para futuras actualizaciones.

La orquestación del pipeline mediante Prefect permitió automatizar todas las etapas críticas: carga de datos, preprocesamiento, optimización de hiperparámetros con Optuna, reentrenamiento con los mejores parámetros y registro de los modelos finales. Este enfoque estructurado hizo posible reproducir el mismo proceso en cualquier momento, mantener el orden en los experimentos y reducir errores manuales. La integración con MLflow en Databricks reforzó esta organización, ya que cada experimento quedó documentado con sus métricas, parámetros, artefactos y modelos, lo cual facilita auditorías, comparaciones y colaboración entre desarrolladores.

El desarrollo de una API con FastAPI y una interfaz de usuario con Streamlit permitió poner el modelo en funcionamiento de manera práctica. La API aplica exactamente el mismo preprocesamiento usado en el entrenamiento para garantizar coherencia en las predicciones, mientras que la interfaz gráfica permite que cualquier usuario ingrese valores y obtenga resultados sin conocer los detalles técnicos del modelo.

Finalmente, se desarrolló la sección correspondiente a Docker, lo que permitió ajustar la configuración del proyecto a las capacidades de cómputo disponibles, realizando las optimizaciones necesarias para asegurar su correcta ejecución bajo las limitaciones de memoria. Debido al tiempo requerido para estas adaptaciones, no fue posible completar la integración con AWS, aunque se considera una etapa viable para implementaciones futuras. Aun así, los resultados obtenidos permitieron cerrar adecuadamente el ciclo del proyecto, avanzando de la fase experimental hacia un despliegue funcional y estable.
